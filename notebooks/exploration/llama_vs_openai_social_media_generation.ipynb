```python
import json
import random
```

```python
def load_jsonl(file_path):
    """Load JSONL file into a list of dictionaries."""
    data = []
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            data.append(json.loads(line.strip()))
    return data

def load_json(file_path):
    """Load JSON file into a list of dictionaries."""
    with open(file_path, 'r', encoding='utf-8') as f:
        return json.load(f)
```

```python
def find_matching_sample(original_post, dataset):
    """Find a sample in dataset that matches the original_post."""
    for sample in dataset:
        if sample.get('original_post') == original_post:
            return sample
    return None

def find_matching_sample_in_original(generated_post, original_dataset):
    """Find the original sample that matches a generated post."""
    for sample in original_dataset:
        # For original dataset, we need to match by text content
        if sample.get('text') == generated_post or sample.get('post') == generated_post:
            return sample
    return None
```

```python
def compare_samples():
    # File paths for all models
    file_paths = {
        'original': 'arabic_datasets/social_media_dataset.json',
        'llama-batched': 'generated_arabic_datasets/llama-batched/arabic_social_media_dataset/by_polishing_posts_generation_filtered.jsonl',
        'openai': 'generated_arabic_datasets/openai/arabic_social_media_dataset/by_polishing_posts_generation_filtered.jsonl',
        'jais-batched': 'generated_arabic_datasets/jais-batched/arabic_social_media_dataset/by_polishing_posts_generation_filtered.jsonl',
        'allam': 'generated_arabic_datasets/allam/arabic_social_media_dataset/by_polishing_posts_generation_filtered.jsonl'
    }
    
    # Load all datasets
    print("Loading datasets...")
    datasets = {}
    
    # Load original dataset (JSON format)
    datasets['original'] = load_json(file_paths['original'])
    print(f"Loaded {len(datasets['original'])} samples from original dataset")
    
    # Load generated datasets (JSONL format)
    for model_name in ['llama-batched', 'openai', 'jais-batched', 'allam']:
        datasets[model_name] = load_jsonl(file_paths[model_name])
        print(f"Loaded {len(datasets[model_name])} samples from {model_name} dataset")
    
    # Randomly select a sample from one of the generated datasets
    generated_models = ['llama-batched', 'openai', 'jais-batched', 'allam']
    source_model = random.choice(generated_models)
    random_sample = random.choice(datasets[source_model])
    original_post = random_sample.get('original_post')
    
    if not original_post:
        print("No original_post found in the selected sample!")
        return
    
    # Find matching samples in all other datasets
    matching_samples = {}
    for model_name in generated_models:
        if model_name != source_model:
            matching_samples[model_name] = find_matching_sample(original_post, datasets[model_name])
    
    # Add the source sample
    matching_samples[source_model] = random_sample
    
    # Display results
    print("\n" + "="*80)
    print("ORIGINAL POST:")
    print("="*80)
    print(original_post[:250])
    
    # Display generations from all models
    model_order = ['openai', 'llama-batched', 'jais-batched', 'allam']
    for model_name in model_order:
        print(f"\n" + "="*80)
        print(f"{model_name.upper()} GENERATED:")
        print("="*80)
        
        sample = matching_samples.get(model_name)
        if sample:
            generated_text = sample.get('generated_post', 'N/A')[:250]
            print(generated_text)
        else:
            print(f"No matching sample found for {model_name}")
    
    print("\n" + "="*80)
```

```python
compare_samples()
```
